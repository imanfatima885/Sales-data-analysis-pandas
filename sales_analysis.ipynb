# Sales Data Analysis using Python & Pandas

```python
import pandas as pd
```

```python
df=pd.read_csv("sales_data.csv")
```

```python
# Display first 5 records
print("\n===== FIRST 5 RECORDS =====")
print(df.head())
```

```python
# Display last 5 records
print("\n===== LAST 5 RECORDS =====")
print(df.tail())
```

```python
# Dataset information
print("\n===== DATASET INFORMATION =====")
df.info()
```

```python
# Total sales amount
total_sales = df["Total Amount"].sum()
print("\nTotal Sales Amount:")
print(total_sales)
```

```python
# Average sales amount
average_sales = df["Total Amount"].mean()
print("\nAverage Sales Amount:")
print(round(average_sales, 2))
```

```python
# Product with highest sales
highest_sale = df.loc[df["Total Amount"].idxmax()]
print("\n===== PRODUCT WITH HIGHEST SALES =====")
print(highest_sale)
```

```python
# Best-selling category
best_category = (
    df.groupby("Category")["Total Amount"]
      .sum()
      .idxmax()
)
print("\nBest Selling Category:")
print(best_category)
```

```python
# Sales greater than 5000
print("\n===== SALES GREATER THAN 5000 =====")
high_sales = df[df["Total Amount"] > 5000]
high_sales
```

```python
# Sort by total amount
sorted_df = df.sort_values(
    by="Total Amount",
    ascending=False
)

print("\n===== SORTED DATASET =====")
sorted_df
```

```python
# Save analyzed dataset
sorted_df.to_csv(
    "sales_data_analyzed.csv",
    index=False
)
print("\nAnalyzed dataset saved successfully.")
```
